```python
df = pd.read_csv('/content/twitter_training.csv', header=None)
df.columns = ['ID', 'Topic', 'Sentiment', 'Tweet']
df.head()
```

```python
print(df.shape)
print(df['Sentiment'].value_counts())
```

```python
df['Sentiment'].value_counts().plot(kind='bar', color=['green','red','blue','gray'])
plt.title('Sentiment Distribution')
plt.xlabel('Sentiment')
plt.ylabel('Count')
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
df['Sentiment'].value_counts().plot(kind='pie', autopct='%1.1f%%')
plt.title('Sentiment Distribution Pie Chart')
plt.ylabel('')
plt.show()
```

```python
df['Tweet'] = df['Tweet'].astype(str)
df['Polarity'] = df['Tweet'].apply(lambda x: TextBlob(x).sentiment.polarity)
df['Subjectivity'] = df['Tweet'].apply(lambda x: TextBlob(x).sentiment.subjectivity)
df.head()
```

```python
df['Polarity'].plot(kind='hist', bins=30, color='skyblue')
plt.title('Polarity Distribution of Tweets')
plt.xlabel('Polarity')
plt.tight_layout()
plt.show()
```

```python
df['Subjectivity'].plot(kind='hist', bins=30, color='orange')
plt.title('Subjectivity Distribution of Tweets')
plt.xlabel('Subjectivity')
plt.tight_layout()
plt.show()
```

```python
df['Topic'].value_counts().head(10).plot(kind='bar', color='purple')
plt.title('Top 10 Topics in Tweets')
plt.xlabel('Topic')
plt.ylabel('Count')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
## Conclusions

1. Most tweets are Positive in sentiment
2. Negative tweets are second most common
3. Polarity shows most tweets are neutral to positive
4. Subjectivity shows tweets are mostly subjective
5. TextBlob successfully analyzed sentiment of Twitter data
```
