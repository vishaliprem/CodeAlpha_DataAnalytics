```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('/content/netflix_titles.csv')
df.head()
```

```python
df.shape
```

```python
df['type'].value_counts().plot(kind='bar', color=['skyblue','orange'])
plt.title('Movies vs TV Shows on Netflix')
plt.xlabel('Type')
plt.ylabel('Count')
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
df['type'].value_counts().plot(kind='pie', autopct='%1.1f%%', colors=['skyblue','orange'])
plt.title('Movies vs TV Shows Distribution')
plt.ylabel('')
plt.show()
```

```python
df['country'].value_counts().head(10).plot(kind='bar', color='green')
plt.title('Top 10 Countries with Most Content')
plt.xlabel('Country')
plt.ylabel('Count')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
df['listed_in'].value_counts().head(10).plot(kind='barh', color='purple')
plt.title('Top 10 Genres on Netflix')
plt.xlabel('Count')
plt.tight_layout()
plt.show()
```

```python
df['year_added'] = pd.to_datetime(df['date_added'], errors='coerce').dt.year
df['year_added'].value_counts().sort_index().plot(kind='line', color='red', marker='o')
plt.title('Content Added to Netflix by Year')
plt.xlabel('Year')
plt.ylabel('Count')
plt.tight_layout()
plt.show()
```

```python
df['director'].value_counts().head(10).plot(kind='barh', color='coral')
plt.title('Top 10 Directors on Netflix')
plt.xlabel('Count')
plt.tight_layout()
plt.show()
```

```python
df['rating'].value_counts().plot(kind='bar', color='teal')
plt.title('Content Rating Distribution on Netflix')
plt.xlabel('Rating')
plt.ylabel('Count')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
## Conclusions

1. Movies are more than TV Shows on Netflix
2. United States has the most content on Netflix
3. International Movies is the most popular genre
4. Netflix added most content between 2018-2020
5. TV-MA is the most common content rating
```
